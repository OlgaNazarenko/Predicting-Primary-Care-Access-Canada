```python
import pandas as pd
import numpy as np
import zipfile
from io import BytesIO
import requests
import io
```

```python
# define the URL of the CSV file
url = 'https://www150.statcan.gc.ca/n1/tbl/csv/13100495-eng.zip'

# use pandas.read_csv() to directly read the CSV file into a datafram
response = requests.get(url)

# create BytesIO buffer from ZIP content
buf1 = BytesIO(response.content)

#extract and read files 
with zipfile.ZipFile(buf1, 'r') as zip_ref:

    # list all files in ZIP
    for name in zip_ref.namelist():
        # find all files ending with .csv
        if name.endswith('.csv'):
            # open and read all files from ZIP
            with zip_ref.open(name) as zd:
                df_satisfaction = pd.read_csv(zd)
            print(df_satisfaction)
        break
```
